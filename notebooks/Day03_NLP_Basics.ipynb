Similarity

```python
nlp=spacy.load("en_core_web_lg")
doc1 = nlp("I like salty fries and hamburgers.")
doc2 = nlp("Fast food tastes very good.")
```

```python
# Similarity of two documents
print(doc1, "<->", doc2, doc1.similarity(doc2))
# Similarity of tokens and spans
french_fries = doc1[2:4]
burgers = doc1[5]
print(french_fries, "<->", burgers, french_fries.similarity(burgers))
```

```python
nlp=spacy.load("en_core_web_lg")
doc1 = nlp("I like salty fries and hamburgers.")
doc2 = nlp("Fast food tastes very good.")
```

```python
lowest_sim=float("inf")
lowest_sim_pair=()
for token1 in doc1:
  if token1.is_punct or token1.is_space:
    continue
  for token2 in doc2:
    if token2.is_punct or token2.is_space:
      continue
    if lowest_sim>token1.similarity(token2):
        lowest_sim=token1.similarity(token2)
        lowest_sim_pair=(token1.text,token2.text)
print(lowest_sim_pair)
```
